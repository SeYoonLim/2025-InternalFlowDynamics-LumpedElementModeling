### S.1 Governing Equation

In the original formulation by Wang et al. (2019), the entire fluid volume was represented by four lumped nodes, each corresponding to a simplified large control volume. In the present study, this approach is refined by subdividing the system into physically significant regions, namely the left and right channels, the compression chamber, and the nozzle. For each subdomain, a channel radius $r_n$ and length $l_n$ are defined, resulting in a more continuous and physically interpretable solution along the axial coordinate. In the present work, the governing equations were derived by modifying the formulation (Wang et al., 2019). The pressure variation inside the inkjet channel is derived from the Navier–Stokes equations for incompressible flow:

$$
\rho \frac{\partial u}{\partial t} + \rho (u \cdot \nabla)u - \mu \nabla^{2} u = -\nabla p + \rho g \qquad 
\tag{S.1}
$$

Where $\rho$ denotes the density, $u$ the velocity, $\mu$ the viscosity, $g$ the gravitational acceleration, $p$ the pressure, $\nabla^{2}$ the Laplacian operator, and $\nabla$ the gradient operator.
The body force term due to gravity can be neglected because the Froude number is sufficiently large, and the convective term can also be ignored since the flow inside the channel is laminar. The velocity $u$ can be expressed as the average velocity corresponding to the cross-sectional area $A_c$, and the volumetric flowrate can be written as $q = A_c \cdot u = \pi r^{2} \cdot u$. Depending on the channel radius $r$ and the channel length $l$, the equation for the pressure difference $\Delta p$ can be simplified as follows:

$$
\Delta p = \Delta P_{L} + \Delta P_{R} = -\frac{\rho l}{\pi r^{2}} \frac{\partial q}{\partial t} + \frac{\mu l}{\pi r^{2}} \nabla^{2} q \qquad 
\tag{S.2}
$$

Therefore, the pressure variation consists of the pressure change due to inertial effects $\Delta P_{L}$ and the pressure change due to viscous effects $\Delta P_{R}$.
The pressure variation caused by inertial effects corresponds to the electromotive force, where $L$ represents the inductance, which is formulated according to Faraday’s law corresponding to the electric current.
Therefore, $\rho l / (\pi r^{2})$ corresponds to the inductance $L$:

$$
L = \frac{\rho l}{\pi r^{2}} \qquad 
\tag{S.3}
$$

The pressure variation caused by viscous effects corresponds to the voltage drop described by Ohm’s law. However, the internal flow does not exhibit a strictly linear relationship. To identify an alternative correlation between viscous forces and flow rate, the concept of momentum thickness is introduced, allowing the flow rate to be interpreted as the result of shear forces derived from the oscillatory boundary layer in Stokes boundary-layer theory. Accordingly, the momentum thickness $\delta$ is defined as follows:

$$
\Delta P_{R} = \mu\left(\frac{U_{\infty}}{\delta}\right)\left(\frac{2\pi r l}{\pi r^{2}}\right) = \left(\frac{2\mu l}{\pi r^{2}}\right) q \qquad 
\tag{S.4}
$$

Where, $U_{\infty}$ denotes the free-stream velocity, $\nu$ the kinematic viscosity, $\omega$ the angular frequency, and $T_c$ the vibration period induced by piezoelectric actuation. The viscous term is then evaluated under the assumption that the shear rate $du/dy$ can be approximated as $U_{\infty}/\delta$:

$$
\delta = \int_{0}^{\infty}\left[\frac{u}{U_{\infty}}\left(1 - \frac{u}{U_{\infty}}\right)\right]dy = \sqrt{\frac{2\nu}{\omega}} = \sqrt{\frac{\mu T_{c}}{\pi \rho}} \qquad 
\tag{S.5}
$$

Therefore, the above group of coefficients corresponds to the resistance $R$, which is expressed as follows:

$$
R = \frac{2\mu l}{\pi r^{3}\delta} = \frac{2l}{r^{3}}\sqrt{\frac{\rho}{\pi \mu T_{C}}} \qquad 
\tag{S.6}
$$

The acoustic pressure wave inside the inkjet, generated by the driving voltage, can be obtained using a circuit model based on acoustic capacitance. Accordingly, the terminal voltage is given by the potential difference between the driving source and the capacitor:

$$
U(t) = U_{d}(t) - U_{\mathrm{cap}}(t) \qquad 
\tag{S.7}
$$

Here, the capacitor voltage is determined by the charge–current relationship as:

$$
U_{\mathrm{cap}}(t) = \frac{\Delta Q(t_{0}) + \int_{t_{0}}^{t} I,dt}{C(t)} = \frac{\Delta V(t)}{C(t)} \qquad 
\tag{S.8}
$$

and can be simplified as follows:

$$
C(t) = \frac{\Delta Q}{\Delta U} = \frac{\Delta V}{\Delta p} = \frac{-V_{0},\frac{V}{K},\Delta p}{\Delta p} = -\frac{\pi r^{2} l}{c^{2}\rho(t)} \qquad 
\tag{S.9}
$$

Where $Q$ denotes the charge, $U$ the voltage, $p$ the pressure, $V$ the volume, $V_{0}$ the initial volume, $c$ the speed of sound, and $K$ the bulk modulus representing the volumetric stiffness. The volumetric variation can be expressed by the relation $K = c^{2}\rho(t)$. Since the bulk change in the inkjet flow channel is generally very small and negligible, the density $\rho(t)$ can be assumed to be constant.
The supply voltage $U_{d}(t)$ represents the driving input that determines the magnitude of the acoustic pressure wave generated by the piezoelectric actuation, which can be estimated considering compressibility. The volumetric variation is expressed as a function of the time-dependent radius $r(t)$:

$$
U_{d}(t) = \frac{\Delta V}{C(t)} = \left(\frac{\Delta p}{\Delta V}\right)\Delta V = -\rho c^{2}\frac{\Delta V}{V} = -\rho c^{2}\frac{2\pi r,\Delta r,l}{\pi r^{2}l} = -\rho c^{2}\left(\frac{2\Delta r(t)}{r}\right) \qquad 
\tag{S.10}
$$

According to the pressure–voltage analogy, the pressure rate can be expressed as the supply voltage variation as:

$$
\frac{d p_{d}}{dt} = -\frac{2\rho c^{2}}{r}\frac{d(\Delta r)}{dt} = -\frac{2\rho c^{2}\alpha}{r}\frac{dU_{A}}{dt} \qquad 
\tag{S.11}
$$

where $\alpha$ is the piezoelectric coefficient defined by $\Delta r = \alpha \Delta U_{A}$ and $\alpha = 0.29 \times 10^{-9}\ \mathrm{m/V}$. Subsequently, the Piezo-pressure coefficient $F$ is determined as a conversion coefficient for calculating the pressure amplitude inside the piezoelectric inkjet generated by the voltage drive $dU_{A}/dt$:

$$
F = -\frac{\rho c^{2}\alpha}{r} \qquad 
\tag{S.12}
$$

By applying Kirchhoff’s Current Law (KCL), the relation can be expressed as:

$$
\dot{p} = -\frac{1}{C}q + \frac{1}{C}U_{p}(t) \qquad 
\tag{S.13}
$$

Similarly, from Kirchhoff’s Voltage Law (KVL), the following relation is obtained:

$$
\dot{q} = \frac{1}{L}p - \frac{R}{L}q + \frac{1}{L}U_{q}(t) \qquad 
\tag{S.14}
$$

By differentiating equation (B.13) and substituting equation (B.14), the following expression is obtained:

$$
L\ddot{q} + R\dot{q} + \frac{1}{C}q = \frac{1}{C}U_{p}(t) + \dot{U}_{q}(t) \qquad 
\tag{S.15}
$$

In the actual system, only the actuator-driven voltage $U_{p}(t)$ is considered as the external input. Accordingly, the dynamic behavior of the system can be expressed as:

$$
L\ddot{q} + R\dot{q} + \frac{1}{C}q = F\frac{dU_{A}}{dt} \qquad 
\tag{S.16}
$$


```python
import math
import torch
import matplotlib.pyplot as plt

torch.set_default_dtype(torch.float64)

# ═════════════ 1. Governing equation ─══════════════════

"""
Governing equation:
    L q̈ + R q̇ + (1/C) q = F (dU_A/dt)

where
    L : inertance
    R : resistance
    C : compliance
    F : electromechanical forcing coefficient
    q : volumetric flow rate
    U_A : applied voltage
"""

partition1, partition2, partition3, partition4 = 40, 40, 40, 40
partition_tot = partition1 + partition2 + partition3 + partition4

# Ink property
sigma = 0.047      # surface tension, sigma [N/m]
mu    = 0.0157     # dynamic viscosity, mu [Pa·s]
rho   = 1111.4     # density, rho [kg/m^3]
alpha = 0.29e-9    # piezoelectric coefficient, alpha [m/V]
c     = 1600.0     # speed of sound, c [m/s]

# Inkjet geometry
r = 235e-6                                                         # r, r_nozz      : [m]
r_nozz = 25e-6
l1, l2, l3, l4 = 0.012, 0.0084, 0.00542303, 0.00057697             # l1, l2, l3, l4 : [m]


T_visc = 2.0 * (l1 + l2 + l3 + l4) / c                             # # T_visc       : [s]
delta_v = torch.sqrt(mu * T_visc / torch.tensor(math.pi) / rho)    # delta_v        : [m]

ri_nozzle = torch.linspace(
    r + (r_nozz - r) / partition4,
    r_nozz,
    partition4,
    dtype=torch.get_default_dtype()
)
ri = torch.cat([r * torch.ones(partition_tot - partition4), ri_nozzle])

li = torch.cat([
    (l1 / partition1) * torch.ones(partition1),
    (l2 / partition2) * torch.ones(partition2),
    (l3 / partition3) * torch.ones(partition3),
    (l4 / partition4) * torch.ones(partition4),
])

cap = (torch.pi * (ri ** 2) * li / (c ** 2) / rho)                 # [m^3/Pa]
cap[-1] = (torch.pi * (r_nozz ** 4) / 3.0 / sigma)                 # [m^3/Pa]
ind = (rho * li / torch.pi / (ri ** 2))                            # [Pa·s^2/m^3]
res = (2.0 * mu * li / torch.pi / delta_v / (ri ** 3))             # [Pa·s/m^3]

N = partition_tot
I_N = torch.eye(N)

L_mat = torch.roll(I_N * (1.0 / ind), -1, dims=1) - I_N * (1.0 / ind)
L_mat[0, -1] = 0.0

C_mat = -torch.roll(I_N * (1.0 / cap), 1, dims=1) + I_N * (1.0 / cap)
C_mat[-1, 0] = 0.0

R_mat = -I_N * (res / ind)

A_mat = L_mat @ C_mat
B_mat = R_mat

B_q_bot = torch.cat([
    torch.zeros(partition1, 1),
    torch.ones(partition2, 1),
    torch.zeros(partition3 + partition4, 1)
], dim=0)

T_ = torch.linspace(0.0, 3e-4, 3002)
dt = (T_[1] - T_[0]).item()
nT = T_.numel()

t1, t2, t3, V_max = 6e-6, 26e-6, 32e-6, 30.0
dVdp = (
    V_max / t1 * (1.0 - (T_ > t1).to(torch.get_default_dtype()))
    - V_max / (t3 - t2) * ((T_ > t2).to(torch.get_default_dtype()) - (T_ > t3).to(torch.get_default_dtype()))
)   # [V/s]

force_scale = (B_q_bot * (c ** 2) * rho * 2.0 * alpha / r)         # [Pa/V]
f_1st = force_scale * dVdp.reshape(1, -1)
f_2nd_bot = L_mat @ f_1st

q_init = torch.zeros(N)
qdot_init = torch.zeros(N)

"""
Original governing equation:
    L q̈ + R q̇ + (1/C) q = F (dU_A/dt)

Matrix form used below:
    q̈ = A q + B q̇ + f(t)
"""
```

### S.2 Lumped element modeling

The governing equation for each partition is expressed as a second-order differential form based on the equivalent RLC circuit analogy:

$$
\ddot{q} = -\frac{R}{L}\dot{q} - \frac{1}{LC}q + \frac{F}{L}\frac{dU_A}{dt} \qquad 
\tag{S.17}
$$

Equation (B.17) describes the dynamic coupling between viscous loss, inertial response, and capacitive compressibility under the influence of the piezoelectric actuation $U_A(t)$.

To extend this into a state-space representation, the system can be expressed as:

$$
\dot{X} = A(t)X(t) + B U(t) \qquad 
\tag{S.18}
$$

where $X(t)$ is the state vector containing both the flowrate and its time derivative:

$$
X =
\left[
\begin{matrix}
q_1\
q_2\
q_3\
q_4\
\dot{q}_1\
\dot{q}_2\
\dot{q}_3\
\dot{q}_4
\end{matrix}
\right]^{T}
\qquad 
\tag{S.19}
$$

The system matrix $A(t)$ represents the coupling between inertial and resistive-capacitive components, defined as:

$$
A(t)=
\begin{bmatrix}
0 & I\\[4pt]
\frac{1}{LC} & \frac{R}{L}
\end{bmatrix}
\qquad 
\tag{S.20}
$$

where $I$ denotes the identity matrix corresponding to the number of partitions.

The input matrix $B$ maps the piezoelectric forcing term to each subdomain as:

$$
B =
\left[
\begin{matrix}
0\
\frac{F}{L}\
0\
0\
0\
0\
0\
0
\end{matrix}
\right]^{T}
\qquad 
\tag{S.21}
$$

and the input vector $U_d(t)$ is defined as the time derivative of the applied voltage waveform:

$$
U_d(t)=\left[\frac{dU_A}{dt},\cdots,\frac{dU_A}{dt}\right]\in \mathbb{R}^{t_{\mathrm{fin}}}
\qquad 
\tag{S.22}
$$

Equation (B.23) defines the partitioned structure of the lumped-parameter model, in which each segment (subdomain) represents a distinct physical region.

$$
\begin{aligned}
R &=
\begin{bmatrix}
-\frac{R_1}{L_1} & 0 & \cdots & 0 \\
0 & -\frac{R_2}{L_2} & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & -\frac{R_n}{L_n}
\end{bmatrix}
\\[6pt]
L &=
\begin{bmatrix}
-\frac{1}{L_1} & 0 & 0 & \cdots & 0 \\
\frac{1}{L_2} & -\frac{1}{L_2} & 0 & \cdots & 0 \\
0 & \ddots & \ddots & \ddots & \vdots \\
\vdots & \ddots & \frac{1}{L_{n-1}} & -\frac{1}{L_{n-1}} & 0 \\
0 & \cdots & 0 & \frac{1}{L_n} & -\frac{1}{L_n}
\end{bmatrix}
\\[6pt]
C &=
\begin{bmatrix}
\frac{1}{C_1} & -\frac{1}{C_1} & 0 & \cdots & 0 \\
0 & \frac{1}{C_2} & -\frac{1}{C_2} & \ddots & \vdots \\
\vdots & \ddots & \ddots & \ddots & 0 \\
0 & \cdots & 0 & \frac{1}{C_{n-1}} & -\frac{1}{C_{n-1}} \\
0 & \cdots & \cdots & 0 & \frac{1}{C_n}
\end{bmatrix}
\end{aligned}
\qquad 
\tag{S.23}
$$

The matrices $R$, $L$, and $C$ respectively characterize the viscous losses, inertial losses, and compressive coupling between adjacent partitions. This matrix formulation systematically describes the interaction of pressure and flowrate across neighboring subdomains, ensuring the axial continuity of the model.

To ensure numerical stability in the time-domain simulation of the lumped-element model, the first-order backward differentiation formula was employed.

The first- and second-order temporal derivatives of the flowrate $q(t)$ at time $t+\Delta t$ are discretized as:

$$
\begin{aligned}
\left.\frac{dq}{dt}\right|_{t+\Delta t} &= \frac{q^{t+\Delta t}-q^{t}}{\Delta t},\\[6pt]
\left.\frac{d^{2}q}{dt^{2}}\right|_{t+\Delta t} &= \frac{\dot{q}^{\,t+\Delta t}-\dot{q}^{\,t}}{\Delta t}
\end{aligned}
\qquad 
\tag{S.24}
$$

By substituting these backward-difference approximations into the governing differential equation, the fully implicit update equation for $q^{t+\Delta t}$ is obtained as:

$$
\begin{aligned}
\left(\frac{1}{\Delta t^{2}}L+\frac{1}{\Delta t}R+C^{-1}\right) q^{t+\Delta t}
&= \frac{L}{\Delta t^{2}}\left(2q^{t}-q^{t-\Delta t}\right)
+ \frac{R}{\Delta t}q^{t}
+ U_A^{\,t+\Delta t}
\end{aligned}
\qquad 
\tag{S.25}
$$

To further improve the numerical stability and temporal accuracy of the simulation, the second-order backward differentiation formula was additionally implemented. Unlike the first-order scheme in Equation (B.25), the second-order method achieves better temporal accuracy while maintaining implicit stability, making it suitable for stiff systems governed by coupled viscous–inertial–compressive dynamics.

The first- and second-order backward differences of the flowrate $q(t)$ at $t+\Delta t$ are expressed as:

$$
\begin{aligned}
\dot{q}^{\,t+\Delta t} &= \frac{3q^{t+\Delta t}-4q^{t}+q^{t-\Delta t}}{2\Delta t},\\[6pt]
\ddot{q}^{\,t+\Delta t} &= \frac{3\dot{q}^{\,t+\Delta t}-4\dot{q}^{\,t}+\dot{q}^{\,t-\Delta t}}{2\Delta t},\\[6pt]
h_{\alpha} &= -\frac{2}{\Delta t}q^{t}+\frac{1}{2\Delta t}q^{t-\Delta t},\\[6pt]
h_{\beta} &= -\frac{2}{\Delta t}\dot{q}^{\,t}+\frac{1}{2\Delta t}\dot{q}^{\,t-\Delta t}
\end{aligned}
\qquad 
\tag{S.26}
$$

By substituting these expressions into the governing differential equation, the fully implicit second-order backward differentiation update equation for $q^{t+\Delta t}$ can be written as:

$$
\begin{aligned}
\left(\frac{9}{4\Delta t^{2}}L-\frac{3}{2\Delta t}R-C^{-1}\right) q^{t+\Delta t}
&=
-L\left(\frac{3}{2\Delta t}h_{\alpha}-h_{\beta}\right)
+Rh_{\alpha}
-LU_A^{\,t+\Delta t}
\end{aligned}
\qquad 
\tag{S.27}
$$

Equation (S.27) represents the second-order implicit BDF formulation, which enhances the numerical stability and reduces truncation error compared with the first-order scheme. This approach is particularly effective for capturing transient responses in highly stiff piezoelectric–fluidic coupled systems.


```python
# ═════════════ 2. BDF-2 solver  ─══════════════════
"""
Solve:
    q''(t) = A q(t) + B q'(t) + f(t)
using BDF-2 recursion.

A, B   : (N, N)
f_bot  : (N, nT)
q0     : (N,)
qdot0  : (N,)
dt     : float

Returns
-------
q_hist    : (N, nT+1)
qdot_hist : (N, nT+1)
"""

A = A_mat
B = B_mat
f_bot = f_2nd_bot
q0 = q_init
qdot0 = qdot_init

N_solver, nT_solver = f_bot.shape
I_solver = torch.eye(N_solver, dtype=A.dtype, device=A.device)

q_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)
qdot_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)

q_hist[:, 0] = q0
qdot_hist[:, 0] = qdot0

f0 = f_bot[:, 0]
qddot0 = A @ q0 + B @ qdot0 + f0
qdot_1 = qdot0 + dt * qddot0
q_1 = q0 + dt * qdot_1

q_hist[:, 1] = q_1
qdot_hist[:, 1] = qdot_1

q_nm2, q_nm1 = q0, q_1
qdot_nm2, qdot_nm1 = qdot0, qdot_1

alpha_bdf2 = 3.0 / (2.0 * dt)
beta_bdf2 = 3.0 / (2.0 * dt)

for i_step in range(2, nT_solver + 1):
    f_i = f_bot[:, i_step - 1]

    const_q = (-4.0 / (2.0 * dt)) * q_nm1 + (1.0 / (2.0 * dt)) * q_nm2
    const_d = (-4.0 / (2.0 * dt)) * qdot_nm1 + (1.0 / (2.0 * dt)) * qdot_nm2

    M_q = (beta_bdf2 * alpha_bdf2) * I_solver - A - alpha_bdf2 * B
    rhs = - (beta_bdf2 * const_q + const_d) + (B @ const_q) - f_i

    q_n = torch.linalg.solve(M_q, rhs)
    qdot_n = alpha_bdf2 * q_n + const_q

    q_hist[:, i_step] = q_n
    qdot_hist[:, i_step] = qdot_n

    q_nm2, q_nm1 = q_nm1, q_n
    qdot_nm2, qdot_nm1 = qdot_nm1, qdot_n

q_sol = q_hist
qdot_sol = qdot_hist

# ═════════════ 3. Post-processing and plot ─══════════════════

i_q = 119
q_plot = q_sol[i_q, 1:].detach().cpu().numpy()              # length nT
dVdp_plot = dVdp.detach().cpu().numpy()                     # length nT

T_plot_us = T_.detach().cpu().numpy() * 1e6                 # [us]
q_um3_per_us = q_plot * 1e12                                # [um^3/us]

# dVdp -> trapezoidal voltage U(t) by time integration
U_plot = torch.zeros_like(T_)
U_plot[1:] = torch.cumsum(0.5 * (dVdp[1:] + dVdp[:-1]) * (T_[1:] - T_[:-1]), dim=0)
U_plot = U_plot.detach().cpu().numpy()                      # [V]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4), constrained_layout=True)

ax1.plot(T_plot_us, q_um3_per_us)
ax1.set_title("Flowrate at nozzle")
ax1.set_xlabel(r"Time ($\mu$s)")
ax1.set_ylabel(r"Volumetric flowrate ($\mu$m$^{3}$/$\mu$s)")
ax1.grid(True)

ax2.plot(T_plot_us, U_plot)
ax2.set_title(r"Driving voltage waveform")
ax2.set_xlabel(r"Time ($\mu$s)")
ax2.set_ylabel(r"$Voltage, U_A$ (V)")
ax2.grid(True)

plt.show()
```

```python
import os
import re
import math
import torch
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

torch.set_default_dtype(torch.float64)

# ============================================================
# 1. Figure setting
# ============================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5), constrained_layout=True)

# ============================================================
# 2. Experimental data: velocity -> volumetric flowrate
#    Plot on ax1
# ============================================================
base_dir = r"D:\Code review\2025-Internal Flow Dynamics-Lumped Element Modeling"
file_name = "Veolcity_profile(Experiment).csv"
file_path = os.path.join(base_dir, file_name)

if not os.path.exists(file_path):
    raise FileNotFoundError(f"CSV file not found: {file_path}")

df = pd.read_csv(file_path)

def extract_v_index(col_name):
    match = re.search(r"(\d+)", col_name)
    if match is None:
        raise ValueError(f"Column name does not contain index: {col_name}")
    return int(match.group(1))

sorted_cols = sorted(df.columns, key=extract_v_index)
df = df[sorted_cols]

time = np.arange(1, df.shape[1] + 1)

all_voltage_labels = [24, 26, 28, 30, 32, 34, 36, 38, 40]

if df.shape[0] != len(all_voltage_labels):
    raise ValueError(
        f"Number of rows ({df.shape[0]}) does not match expected voltage cases ({len(all_voltage_labels)})."
    )

target_voltages = [24, 32, 40]
target_indices = [all_voltage_labels.index(v) for v in target_voltages]

# ============================================================
# Q = U * A
# [m^3/s] -> [um^3/us] : multiply by 1e12
# ============================================================
nozzle_radius = 25e-6
nozzle_area = np.pi * nozzle_radius**2
m3s_to_um3us = 1e12

for idx, voltage in zip(target_indices, target_voltages):
    velocity = df.iloc[idx].to_numpy(dtype=float)             # [m/s]
    flowrate_m3s = velocity * nozzle_area                     # [m^3/s]
    flowrate_um3us = flowrate_m3s * m3s_to_um3us             # [um^3/us]

    ax1.plot(time, flowrate_um3us, linewidth=2.0, label=f"{voltage} V")

ax1.set_title("Experimental volumetric flowrate (EG 100)")
ax1.set_xlabel(r"Time ($\mu$s)")
ax1.set_ylabel(r"Volumetric flowrate ($\mu$m$^{3}$/$\mu$s)")
ax1.grid(True)
ax1.legend(title="Driving voltage")

# ============================================================
# 3. Numerical model: EG 100
#    Plot on ax2
# ============================================================
drivingvoltage = [
    {'name': '24 V', 'V_max': 24},
    {'name': '32 V', 'V_max': 32},
    {'name': '40 V', 'V_max': 40},
]

ink = {
    'name': 'EG 100',
    'sigma': 0.0474,
    'mu': 0.0157,
    'rho': 1114.00,
    'alpha': 0.29e-9,
    'c': 1688.00
}

# ═════════════ 1. Governing equation ─══════════════════
"""
Governing equation:
    L q̈ + R q̇ + (1/C) q = F (dU_A/dt)

where
    L : inertance
    R : resistance
    C : compliance
    F : electromechanical forcing coefficient
    q : volumetric flow rate
    U_A : applied voltage
"""

partition1, partition2, partition3, partition4 = 40, 40, 40, 40
partition_tot = partition1 + partition2 + partition3 + partition4

sigma = ink['sigma']   # [N/m]
mu = ink['mu']         # [Pa·s]
rho = ink['rho']       # [kg/m^3]
alpha = ink['alpha']   # [m/V]
c = ink['c']           # [m/s]

r = 235e-6
r_nozz = 25e-6
l1, l2, l3, l4 = 0.012, 0.0084, 0.00542303, 0.00057697

T_ = torch.linspace(0.0, 3e-4, 3002)
dt = (T_[1] - T_[0]).item()
nT = T_.numel()
T_plot_us = T_.detach().cpu().numpy() * 1e6

t1, t2, t3 = 6e-6, 26e-6, 32e-6

T_visc = 2.0 * (l1 + l2 + l3 + l4) / c
delta_v = torch.sqrt(mu * T_visc / torch.tensor(math.pi) / rho)

ri_nozzle = torch.linspace(
    r + (r_nozz - r) / partition4,
    r_nozz,
    partition4,
    dtype=torch.get_default_dtype()
)
ri = torch.cat([r * torch.ones(partition_tot - partition4), ri_nozzle])

li = torch.cat([
    (l1 / partition1) * torch.ones(partition1),
    (l2 / partition2) * torch.ones(partition2),
    (l3 / partition3) * torch.ones(partition3),
    (l4 / partition4) * torch.ones(partition4),
])

cap = (torch.pi * (ri ** 2) * li / (c ** 2) / rho)
cap[-1] = (torch.pi * (r_nozz ** 4) / 3.0 / sigma)
ind = (rho * li / torch.pi / (ri ** 2))
res = (2.0 * mu * li / torch.pi / delta_v / (ri ** 3))

N = partition_tot
I_N = torch.eye(N)

L_mat = torch.roll(I_N * (1.0 / ind), -1, dims=1) - I_N * (1.0 / ind)
L_mat[0, -1] = 0.0

C_mat = -torch.roll(I_N * (1.0 / cap), 1, dims=1) + I_N * (1.0 / cap)
C_mat[-1, 0] = 0.0

R_mat = -I_N * (res / ind)

A_mat = L_mat @ C_mat
B_mat = R_mat

B_q_bot = torch.cat([
    torch.zeros(partition1, 1),
    torch.ones(partition2, 1),
    torch.zeros(partition3 + partition4, 1)
], dim=0)

force_scale = (B_q_bot * (c ** 2) * rho * 2.0 * alpha / r)

for dv in drivingvoltage:
    name = dv['name']
    V_max = dv['V_max']

    dVdp = (
        V_max / t1 * (1.0 - (T_ > t1).to(torch.get_default_dtype()))
        - V_max / (t3 - t2) * ((T_ > t2).to(torch.get_default_dtype()) - (T_ > t3).to(torch.get_default_dtype()))
    )

    f_1st = force_scale * dVdp.reshape(1, -1)
    f_2nd_bot = L_mat @ f_1st

    q_init = torch.zeros(N)
    qdot_init = torch.zeros(N)

    A = A_mat
    B = B_mat
    f_bot = f_2nd_bot
    q0 = q_init
    qdot0 = qdot_init

    N_solver, nT_solver = f_bot.shape
    I_solver = torch.eye(N_solver, dtype=A.dtype, device=A.device)

    q_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)
    qdot_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)

    q_hist[:, 0] = q0
    qdot_hist[:, 0] = qdot0

    f0 = f_bot[:, 0]
    qddot0 = A @ q0 + B @ qdot0 + f0
    qdot_1 = qdot0 + dt * qddot0
    q_1 = q0 + dt * qdot_1

    q_hist[:, 1] = q_1
    qdot_hist[:, 1] = qdot_1

    q_nm2, q_nm1 = q0, q_1
    qdot_nm2, qdot_nm1 = qdot0, qdot_1

    alpha_bdf2 = 3.0 / (2.0 * dt)
    beta_bdf2 = 3.0 / (2.0 * dt)

    for i_step in range(2, nT_solver + 1):
        f_i = f_bot[:, i_step - 1]

        const_q = (-4.0 / (2.0 * dt)) * q_nm1 + (1.0 / (2.0 * dt)) * q_nm2
        const_d = (-4.0 / (2.0 * dt)) * qdot_nm1 + (1.0 / (2.0 * dt)) * qdot_nm2

        M_q = (beta_bdf2 * alpha_bdf2) * I_solver - A - alpha_bdf2 * B
        rhs = - (beta_bdf2 * const_q + const_d) + (B @ const_q) - f_i

        q_n = torch.linalg.solve(M_q, rhs)
        qdot_n = alpha_bdf2 * q_n + const_q

        q_hist[:, i_step] = q_n
        qdot_hist[:, i_step] = qdot_n

        q_nm2, q_nm1 = q_nm1, q_n
        qdot_nm2, qdot_nm1 = qdot_nm1, qdot_n

    q_sol = q_hist

    i_q = 119
    q_plot = q_sol[i_q, 1:].detach().cpu().numpy()

    T_plot_us_local = T_plot_us[:q_plot.shape[0]]
    q_um3_per_us = q_plot * 1e12 / 3.0

    ax2.plot(T_plot_us_local, q_um3_per_us, linewidth=2.0, label=name)

ax2.set_title("Analytical volumetric flowrate (EG 100)")
ax2.set_xlabel(r"Time ($\mu$s)")
ax2.set_ylabel(r"Volumetric flowrate ($\mu$m$^{3}$/$\mu$s)")
ax2.grid(True)
ax2.legend(title="Driving voltage")

plt.show()
```

### Result 3. Volumetric flowrate at nozzle for EG 100 under different driving voltages

```python
import math
import torch
import matplotlib.pyplot as plt

torch.set_default_dtype(torch.float64)

drivingvoltage = [
    {'name': '20 V', 'V_max': 20},
    {'name': '24 V', 'V_max': 24},
    {'name': '28 V', 'V_max': 28},
    {'name': '32 V', 'V_max': 32},
    {'name': '36 V', 'V_max': 36},
    {'name': '40 V', 'V_max': 40},
]

# Fixed ink property: EG 100
ink = {
    'name': 'EG 100',
    'sigma': 0.0474,
    'mu': 0.0157,
    'rho': 1114.00,
    'alpha': 0.29e-9,
    'c': 1688.00
}

# ═════════════ 1. Governing equation ─══════════════════
"""
Governing equation:
    L q̈ + R q̇ + (1/C) q = F (dU_A/dt)

where
    L : inertance
    R : resistance
    C : compliance
    F : electromechanical forcing coefficient
    q : volumetric flow rate
    U_A : applied voltage
"""

partition1, partition2, partition3, partition4 = 40, 40, 40, 40
partition_tot = partition1 + partition2 + partition3 + partition4

# Fixed ink property
sigma = ink['sigma']   # [N/m]
mu = ink['mu']         # [Pa·s]
rho = ink['rho']       # [kg/m^3]
alpha = ink['alpha']   # [m/V]
c = ink['c']           # [m/s]

# Inkjet geometry
r = 235e-6
r_nozz = 25e-6
l1, l2, l3, l4 = 0.012, 0.0084, 0.00542303, 0.00057697

# Common time setting
T_ = torch.linspace(0.0, 3e-4, 3002)
dt = (T_[1] - T_[0]).item()
nT = T_.numel()
T_plot_us = T_.detach().cpu().numpy() * 1e6  # [us]

# Common trapezoid timing
t1, t2, t3 = 6e-6, 26e-6, 32e-6

# Geometry-dependent matrices for EG 100
T_visc = 2.0 * (l1 + l2 + l3 + l4) / c
delta_v = torch.sqrt(mu * T_visc / torch.tensor(math.pi) / rho)

ri_nozzle = torch.linspace(
    r + (r_nozz - r) / partition4,
    r_nozz,
    partition4,
    dtype=torch.get_default_dtype()
)
ri = torch.cat([r * torch.ones(partition_tot - partition4), ri_nozzle])

li = torch.cat([
    (l1 / partition1) * torch.ones(partition1),
    (l2 / partition2) * torch.ones(partition2),
    (l3 / partition3) * torch.ones(partition3),
    (l4 / partition4) * torch.ones(partition4),
])

cap = (torch.pi * (ri ** 2) * li / (c ** 2) / rho)     # [m^3/Pa]
cap[-1] = (torch.pi * (r_nozz ** 4) / 3.0 / sigma)     # [m^3/Pa]
ind = (rho * li / torch.pi / (ri ** 2))                # [Pa·s^2/m^3]
res = (2.0 * mu * li / torch.pi / delta_v / (ri ** 3)) # [Pa·s/m^3]

N = partition_tot
I_N = torch.eye(N)

L_mat = torch.roll(I_N * (1.0 / ind), -1, dims=1) - I_N * (1.0 / ind)
L_mat[0, -1] = 0.0

C_mat = -torch.roll(I_N * (1.0 / cap), 1, dims=1) + I_N * (1.0 / cap)
C_mat[-1, 0] = 0.0

R_mat = -I_N * (res / ind)

A_mat = L_mat @ C_mat
B_mat = R_mat

B_q_bot = torch.cat([
    torch.zeros(partition1, 1),
    torch.ones(partition2, 1),
    torch.zeros(partition3 + partition4, 1)
], dim=0)

force_scale = (B_q_bot * (c ** 2) * rho * 2.0 * alpha / r)  # [Pa/V]

# Plot setting
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5), constrained_layout=True)

# ═════════════ 2. Loop over driving voltage ─══════════════════
for dv in drivingvoltage:
    name = dv['name']
    V_max = dv['V_max']

    # dU_A/dt
    dVdp = (
        V_max / t1 * (1.0 - (T_ > t1).to(torch.get_default_dtype()))
        - V_max / (t3 - t2) * ((T_ > t2).to(torch.get_default_dtype()) - (T_ > t3).to(torch.get_default_dtype()))
    )   # [V/s]

    # Trapezoidal voltage U(t)
    U_plot = torch.zeros_like(T_)
    U_plot[1:] = torch.cumsum(
        0.5 * (dVdp[1:] + dVdp[:-1]) * (T_[1:] - T_[:-1]),
        dim=0
    )
    U_plot_np = U_plot.detach().cpu().numpy()  # [V]

    # Plot voltage waveform on ax1
    ax1.plot(T_plot_us, U_plot_np, label=name)

    # Forcing term
    f_1st = force_scale * dVdp.reshape(1, -1)
    f_2nd_bot = L_mat @ f_1st

    q_init = torch.zeros(N)
    qdot_init = torch.zeros(N)

    """
    Original governing equation:
        L q̈ + R q̇ + (1/C) q = F (dU_A/dt)

    Matrix form used below:
        q̈ = A q + B q̇ + f(t)
    """

    # ═════════════ 3. BDF-2 solver ─══════════════════
    A = A_mat
    B = B_mat
    f_bot = f_2nd_bot
    q0 = q_init
    qdot0 = qdot_init

    N_solver, nT_solver = f_bot.shape
    I_solver = torch.eye(N_solver, dtype=A.dtype, device=A.device)

    q_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)
    qdot_hist = torch.zeros((N_solver, nT_solver + 1), dtype=A.dtype, device=A.device)

    q_hist[:, 0] = q0
    qdot_hist[:, 0] = qdot0

    f0 = f_bot[:, 0]
    qddot0 = A @ q0 + B @ qdot0 + f0
    qdot_1 = qdot0 + dt * qddot0
    q_1 = q0 + dt * qdot_1

    q_hist[:, 1] = q_1
    qdot_hist[:, 1] = qdot_1

    q_nm2, q_nm1 = q0, q_1
    qdot_nm2, qdot_nm1 = qdot0, qdot_1

    alpha_bdf2 = 3.0 / (2.0 * dt)
    beta_bdf2 = 3.0 / (2.0 * dt)

    for i_step in range(2, nT_solver + 1):
        f_i = f_bot[:, i_step - 1]

        const_q = (-4.0 / (2.0 * dt)) * q_nm1 + (1.0 / (2.0 * dt)) * q_nm2
        const_d = (-4.0 / (2.0 * dt)) * qdot_nm1 + (1.0 / (2.0 * dt)) * qdot_nm2

        M_q = (beta_bdf2 * alpha_bdf2) * I_solver - A - alpha_bdf2 * B
        rhs = - (beta_bdf2 * const_q + const_d) + (B @ const_q) - f_i

        q_n = torch.linalg.solve(M_q, rhs)
        qdot_n = alpha_bdf2 * q_n + const_q

        q_hist[:, i_step] = q_n
        qdot_hist[:, i_step] = qdot_n

        q_nm2, q_nm1 = q_nm1, q_n
        qdot_nm2, qdot_nm1 = qdot_nm1, qdot_n

    q_sol = q_hist

    # ═════════════ 4. Post-processing ─══════════════════
    i_q = 119
    q_plot = q_sol[i_q, 1:].detach().cpu().numpy()

    # Length safety
    T_plot_us_local = T_plot_us[:q_plot.shape[0]]
    q_um3_per_us = q_plot * 1e12 / 2.0  # [um^3/us]

    # Plot flowrate on ax2
    ax2.plot(T_plot_us_local, q_um3_per_us, label=name)

# ═════════════ 5. Final plotting ─══════════════════

# ax1: trapezoidal voltage waveforms
ax1.set_title("Trapezoidal driving voltage waveforms")
ax1.set_xlabel(r"Time ($\mu$s)")
ax1.set_ylabel(r"$Voltage, U_A$ (V)")
ax1.grid(True)
ax1.legend()

# ax2: flowrate for EG 100 at different Vmax
ax2.set_title("Volumetric flowrate at nozzle for EG 100 under different driving voltages")
ax2.set_xlabel(r"Time ($\mu$s)")
ax2.set_ylabel(r"Volumetric flowrate ($\mu$m$^{3}$/$\mu$s)")
ax2.grid(True)
ax2.legend()

plt.show()
```
